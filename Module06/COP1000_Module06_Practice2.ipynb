```python
scores = []

s1 = int(input("Enter your first score: "))
s2 = int(input("Enter your second score: "))
s3 = int(input("Enter your third score: "))
s4 = int(input("Enter your fourth score: "))
s5 = int(input("Enter your fifth score: "))

scores.extend([s1, s2, s3, s4, s5])
print(scores)

print(sum(scores))
print(len(scores))

total = sum(scores)
amount = len(scores)

average = float(total / amount)
print(f"{average:.2f}")

print(max(scores))
print(min(scores))
```
