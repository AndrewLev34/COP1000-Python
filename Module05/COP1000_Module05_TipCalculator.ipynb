```python
#Ask user for total bill and desired tip percentage
#use a calculate_tip and calculate_total function

def display_title():

    print("Tip Calculator")
    print("--------------\n")


display_title()

def calculate_tip(bill, tip_percent):
    return bill * tip_percent

def calculate_total(bill, tip):
    return bill + tip


bill_cost = float(input("Enter bill cost:"))
tip_percentage = float(input("Enter Desired tip percentage:")) / 100

tip_amount = calculate_tip(bill_cost, tip_percentage)
total = calculate_total(bill_cost, tip_amount)

print()
print(f"Before tip: ${bill_cost}")
print(f"Your tip will be: ${tip_amount:.2f}")
print(f"Your total will be: ${total:.2f}")
```
