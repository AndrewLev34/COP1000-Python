```python
def add_numbers(number1, number2):
    return number1 + number2

num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))

result = add_numbers(num1, num2)
print(f"Sum:{result} ")
```

```python
def calculate_area(length, width):
    return length * width

input_1 = int(input("Enter the length: "))
input_2 = int(input("Enter the width:"))

result = calculate_area(input_1, input_2)
print(f"Area: {result}")
```

```python
def fahrenheit_to_celsius(fahrenheit):
    return (fahrenheit - 32) * 5/9

temp_in_f = float(input("Enter the temperature in fahrenheit: "))

result = fahrenheit_to_celsius(temp_in_f)
print(f"Temperature in Celsius: {result:.1f}")
```

```python
def get_letter_grade(score):
    if score >= 90:
        letter = "A"
    elif 89 >= score >= 80:
        letter = "B"
    elif 79 >= score >= 70:
        letter = "C"
    elif 69 >= score >= 60:
        letter = "D"
    elif score <= 59:
        letter = "F"

    return letter

number = int(input("Enter your score: "))

result = get_letter_grade(number)
print(f"You got an {result}.")
```

```python
def calculate_total(price, quantity):
    return price * quantity

item_name = input("Enter Item name: ")
item_price = float(input("Enter Price: "))
item_quantity = int(input("Enter Quantity: "))

result = calculate_total(item_price, item_quantity)

def display_total(item, total):

    print(f"Item: {item}")
    print(f"Total: ${total:.2f}")

display_total(item_name, result)
```
