```python
# add three contacts as well as their phone and emails

contacts = {
    "John": {"phone": "904 123 1234", "email": "john@gmail.com"},
    "Bob": {"phone": "904 456 4567", "email": "bob@gmail.com"},
    "Jordan": {"phone": "352 789 7890", "email": "jordan@gmail.com"}
}

print("Contact Book\n")

print("Available Contacts")

for name in contacts.keys():
    print(f"{name}")

print()

user_request = input("Enter Contact Name: ")

if user_request in contacts:
    print(f"{name}'s Phone Number is {contacts[user_request]["phone"]}\n{name}'s Email is {contacts[user_request]["email"]}")
else:
    print("Contact Not Found")
```
