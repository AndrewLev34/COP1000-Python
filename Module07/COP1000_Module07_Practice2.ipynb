```python
python_students = {"Alex", "Jordan", "Taylor", "Morgan"}
web_students = {"Jordan", "Morgan", "Casey", "Riley"}

print(f"Students in both courses: {python_students & web_students}")
print(f"Students in either course: {python_students | web_students}")
print(f"Students in Pyton but not Web: {python_students - web_students}")
```
