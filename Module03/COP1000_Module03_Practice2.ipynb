```python
age = int(input("enter age:"))
answer = input("Do you have a valid ID? (yes/no):").lower()

if age >= 18 and answer == "yes":
  print("Entry Permitted")
elif age >= 18 and answer == "no":
  print("Entry Denied")
elif age <= 18 and answer == "no":
  print("Enrty Denied")
elif age <= 18 and answer == "yes":
  print("Entry Denied")
```
