```python
age = 19
credits_completed = 30

age >= 18
credits_completed >= 24

print(f"Age requirement met: {age >= 18}")
print(f"Credit requirement met: {credits_completed >= 24}")
print(f"All requirements met: {age >= 18 and credits_completed >= 24}")
```
