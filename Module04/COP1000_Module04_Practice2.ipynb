```python
password = "python123"

user_entry = input("Enter password: ")

while user_entry != password:
  print("Incorrect password. Try again.")
  user_entry = input("Enter password:")

if user_entry == password:
  print("Access Granted.")
```

```python
total = 0

user_entry = int(input("Enter a number: "))

while user_entry != 0:
  total += user_entry
  user_entry = int(input("Enter a number: "))

if user_entry == 0:
  print()
  print(f"Total: {total} ")
```

```python
password = "python"

attempts = 0
max_attempts = 3

while attempts < max_attempts:
  user_entry = input("Enter password: ")

  if user_entry == password:
    print("Correct!")
    break

  else:
    attempts += 1
    print(f"Incorrect. {max_attempts - attempts} remaining.")

if attempts == max_attempts:
  print("No attempts remaining.")
```
