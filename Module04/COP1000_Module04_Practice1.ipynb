```python
countdown = 10

while countdown > 0:
  print(countdown)
  countdown -= 1

print("Blastoff!")
```

```python
number = int(input("Enter a Number: "))

for i in range(1,11):
  result = number * i
  print(f"{number} x {i} = {result}")
```

```python
total = 0

for i in range(1,11):
  total += i
  print(total)
```

```python
total = 0

for i in range(1,6):
  score = float(input(f"Enter Test Score {i}: "))
  total += score


print(f"Total: {total}")
print(f"Average: {total / 5}")
```
