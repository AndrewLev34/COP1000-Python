```python
import csv

print("Activity Data Logger")

act1 = input("Enter the first activity and time: ")
act2 = input("Enter the second activity and time: ")
act3 = input("Enter the third activity and time: ")

def parse_input(user_input):
    if "," in user_input:
        parts = user_input.split(",", 1)
        return [parts[0].strip(), parts[1].strip()]
    return [user_input.strip(), ""]

with open("activity_data.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Activity", "Time"])
    writer.writerow(parse_input(act1))
    writer.writerow(parse_input(act2))
    writer.writerow(parse_input(act3))

with open("activity_data.csv", "r", newline="") as file:
    reader = csv.reader(file)
    for row in reader:
        print("{} - {}".format(row[0],row[1])  )
```
