```python
import csv

with open("students.csv", "r") as file:
    reader = csv.reader(file)
    rows = list(reader)

if len(rows[0]) > 3:
    rows[0] = rows[0][:3]
for i in range(1, len(rows)):
    rows[i] = rows[i][:3]

rows[0].append("Credits Completed")

completed_credits = ["26", "38", "44"]
for i in range(1, len(rows)):
    rows[i].append(completed_credits[i - 1])

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(rows)

with open("students.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print("{} - {} - {} - {}".format(row[0], row[1], row[2], row[3]))
```

```python
import csv

with open("students.csv", "r") as file:
    reader = csv.reader(file)
    rows = list(reader)

cleaned_rows = []

header = [rows[0][0], rows[0][1], "Credits Completed"]
cleaned_rows.append(header)

completed_credits = ["26", "38", "44"]
for i in range(1, len(rows)):
    name = rows[i][0]
    major = rows[i][1]
    completed_val = completed_credits[i - 1]
    cleaned_rows.append([name, major, completed_val])

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(cleaned_rows)

with open("students.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(" - ".join(row))
```
