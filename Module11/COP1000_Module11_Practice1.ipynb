```python
import math

print(f"Pi: {math.pi}")
print(f"Square Root of 81: {math.sqrt(81)}")
print(f"5 Squared: {math.pow(5,2)}")
```

```python
radius = int(input(f"Enter the radius of a circle: "))
area = math.pi * math.pow(radius, 2)

print(f"The area of your circle is {area:.2f} square units.")
```
