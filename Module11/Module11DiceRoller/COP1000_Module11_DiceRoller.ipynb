```python
#Remember to upload the dice_tools.py file before calling it

import dice_tools

sides = int(input("How many sides are on your die?: "))
dicenum = int(input("How many dice should be rolled?: "))
print()

total = 0

for i in range (dicenum):
    print(f"Roll {i+1}: {dice_tools.roll_die(sides)}")
    roll = dice_tools.roll_die(sides)
    total += roll
print()

print(f"Total: {total}")



```
