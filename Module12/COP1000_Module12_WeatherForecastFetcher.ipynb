```python
try:
    import requests
    print("Requests Library Imported\n")
except ImportError:
    print("Requests Library Not Imported\n")


url = "https://geocoding-api.open-meteo.com/v1/search"

#build in a status check

user_city = input("Enter a City: ")
user_state = input("Enter a State: ")

params = {
    "name": user_city,
    "admin1": user_state,
    "language": "en",
    "format": "json"
}

response = requests.get(url, params=params)

if response.status_code == 200:
    data = response.json()
    print(data)
else:
    print("Error:", response.status_code)

print()

location = data["results"][0]

print(f"Name: {location['name']}")
print(f"State: {location['admin1']}")
print(f"Country: {location['country']}")
print(f"Latitude: {location['latitude']}")
print(f"Longitude: {location['longitude']}")

weather_url = "https://api.open-meteo.com/v1/forecast"

weather_params = {
    "latitude": location["latitude"],
    "longitude": location["longitude"],
    "current": "temperature_2m,wind_speed_10m",
    "temperature_unit": "fahrenheit",
    "wind_speed_unit": "mph"
}


weather = requests.get(weather_url, params=weather_params)

if weather.status_code == 200:
    print(f"Weather data retrieved successfully.\n\n{weather.json()}")
else:
    print("Failed to retrieve weather data.")


current_weather = weather.json()

print(f"\nCurrent Weather in {user_city}, {user_state}\n")
print(f"Temperature: {current_weather['current']['temperature_2m']}°F")
print(f"Wind Speed: {current_weather['current']['wind_speed_10m']} mph")
```
