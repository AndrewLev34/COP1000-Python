```python
weather = {
    "city": "Jacksonville",
    "current": {
        "temperature": 82,
        "humidity": 68,
        "wind_speed": 9
    }
}

for key, value in weather.items():
    if isinstance(value, dict):
        print(f"{key}:")
        for sub_key, sub_value in value.items():
            print(f"{sub_key}: {sub_value}")
    else: print(f"{key}: {value}")
```

```python
forecast = {
    "city": "Jacksonville",
    "temperatures": [82, 84, 86, 83, 79]
}

for key, value in forecast.items():
    if isinstance(value, list):
        print(f"{key}:")
        for item in value:
            print(item)
    else: print(f"{key}: {value}")
print()

print(f"{forecast["temperatures"][0]}")
print(f"{forecast["temperatures"][-1]}")
print(f"{min(forecast["temperatures"])}")
print(f"{max(forecast["temperatures"])}")
```

```python
data = {
    "cities": [
        {"name": "Jacksonville", "temperature": 82},
        {"name": "Orlando", "temperature": 87},
        {"name": "Tallahassee", "temperature": 84}
    ]
}

for city in data["cities"]:
    print(f"{city['name']}: {city['temperature']}")
```
