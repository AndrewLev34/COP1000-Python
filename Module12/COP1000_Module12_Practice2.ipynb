```python
try:
    import requests
    print("Requests library is loaded.")
except ImportError:
    print("Requests library is not loaded.")
```

```python
url = "https://geocoding-api.open-meteo.com/v1/search"

params = {
    "name": "Jacksonville",
    "count": 1,
    "language": "en",
    "format": "json"
}

response = requests.get(url, params=params)

response.status_code
```

```python
data = response.json()

print(data)

type(data)
```

```python
location = data["results"][0]

print(f"Name: {location["name"]}")
print(f"State: {location["admin1"]}")
print(f"Country: {location["country"]}")
print(f"Latitude: {location["latitude"]}")
print(f"Longitude: {location["longitude"]}")
```

```python
weather_url = "https://api.open-meteo.com/v1/forecast"

weather_params = {
    "latitude": location["latitude"],
    "longitude": location["longitude"],
    "current": "temperature_2m,wind_speed_10m",
    "temperature_unit": "fahrenheit",
    "wind_speed_unit": "mph"
}

weather = requests.get(weather_url, weather_params)

if weather.status_code == 200:
    print(f"""Weather data retrieved successfully.\n
{weather.json()}""")
else:
    print("Failed to retrieve weather data.")
```

```python
current_weather = weather.json()

print("Current Weather")

print(f"Temperature: {current_weather["current"]["temperature_2m"]}")
print(f"Wind Speed: {current_weather["current"]["wind_speed_10m"]}")
```
