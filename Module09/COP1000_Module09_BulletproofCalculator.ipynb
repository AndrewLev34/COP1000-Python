```python
#Addition: +
#Subtraction: -
#Multiplication: *
#Division: /

print("Bulletproof Calculator\n")

try:
    user_num1 = int(input("Enter a number: "))
    user_num2 = int(input("Enter another number: "))

    user_action = input("+, -, *, or /? ")

    if user_action == "+":
        print(user_num1 + user_num2)
    elif user_action == "-":
        print(user_num1 - user_num2)
    elif user_action == "*":
        print(user_num1 * user_num2)
    elif user_action == "/":
        result = user_num1 / user_num2
        print(f"{result:.2f}")
    else:
        print("Invalid action. Please Enter +, -, *, or /.")
except ValueError:
    print("Invalid. Enter a Whole Number.")
except ZeroDivisionError:
    print("Invalid. Cannot divide by zero.")
```
