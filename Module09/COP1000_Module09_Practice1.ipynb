```python
user_input = input("Enter your age: ")
try:
    age = int(user_input)
    print(f"You are {age} years old.")
except ValueError as error:
    print("Input Rejected:", type(error).__name__)
    print("*Enter a valid whole number*")
```

```python
num1 = input("Enter a number: ")
num2 = input("Enter a second number: ")

try:
    val1 = int(num1)
    val2 = int(num2)
    answer = val1 / val2
    print(f"The answer is {answer}")
except ValueError as error:
    print("Input Rejected:", type(error).__name__)
except ZeroDivisionError as error:
    print("You cannot divide by zero")
```

```python
grade = input("Enter a grade between 0-100: ")

try:
    value = int(grade)
    if value < 0 or value > 100:
        raise ValueError("Grade out of range")
    print(f"{value} is a valid grade.")
except ValueError as error:
    print("Input Rejected:", type(error).__name__)
    print("*Enter a valid whole number between 0 and 100*")
```

```python
while True:
    age = input("Enter your age: ")

    try:
        age_num = int(age)
        if age_num < 1:
            raise ValueError("Age out of range")
        print(f"{age_num} is a valid age.")
        break

    except ValueError as error:
        print("Input Rejected:", type(error).__name__)
        if "Age out of range" in str(error):
            print("*Age cannot be 0 or less*")
        else:
            print("*Enter a whole number*")
```
