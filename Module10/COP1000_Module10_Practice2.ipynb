```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def calculate_area(self):
        return self.length * self.width

rectangle1 = Rectangle(15, 7)
area = rectangle1.calculate_area()
print(f"The area of the rectangle is: {area}")

rectangle2 = Rectangle(7, 5)
area = rectangle2.calculate_area()
print(f"The area of the rectangle is: {area}")
#
```

```python
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

holder = BankAccount("Mark", 1000)

print(f"Account holder: {holder.account_holder}")
print(f"Balance: {holder.balance}")

holder.deposit(500)
print(f"Balance after deposit: {holder.balance}")
```

```python
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds")
        else:
            self.balance -= amount
            print(f"Withdrawal successful. New balance: {self.balance}")

holder = BankAccount("Mark", 1000)

print(f"Account holder: {holder.account_holder}")
print(f"Balance: {holder.balance}")

holder.withdraw(1100)
```

```python
class Book:
    def __init__(self, title, author, availability):
        self.title = title
        self.author = author
        self.availability = availability

    def check_availability(self):
        if self.availability:
            print(f"{self.title} by {self.author} is available.")
        else:
            print(f"{self.title} by {self.author} is not available.")

    def rent_book(self):
        if self.availability:
            self.availability = False
            print(f"{self.title} by {self.author} has been rented.")
        else:
            print(f"{self.title} by {self.author} is already rented and not available.")

    def return_book(self):
        if not self.availability:
            self.availability = True
            print(f"{self.title} by {self.author} has been returned.")
        else:
            print(f"{self.title} by {self.author} is now available.")

book1 = Book("The Hunger Games", "Suzanne Collins", True)

print(f"Title: {book1.title}")
print(f"Author: {book1.author}")
book1.check_availability()
print()

book1.rent_book()
book1.check_availability()
print()

book1.return_book()
book1.check_availability()
```

```python
class Book:
    def __init__(self, title, author, availability):
        self.title = title
        self.author = author
        self.availability = availability

    def check_availability(self):
        if self.availability:
            print(f"{self.title} by {self.author} is available.")
        else:
            print(f"{self.title} by {self.author} is not available.")

    def rent_book(self):
        if self.availability:
            self.availability = False
            print(f"{self.title} by {self.author} has been rented.")
        else:
            print(f"{self.title} by {self.author} is already rented and not available.")

    def return_book(self):
        if not self.availability:
            self.availability = True
            print(f"{self.title} by {self.author} has been returned.")
        else:
            print(f"{self.title} by {self.author} is now available.")

book1 = Book("The Hunger Games", "Suzanne Collins", True)
book2 = Book("The Great Gatsby", "F. Scott Fitzgerald", False)
book3 = Book("A Storm Too Soon", "Michael J. Tougias", True)

books = [book1, book2, book3]

for book in books:
    print(f"Title: {book.title}")
    print(f"Author: {book.author}")
    book.check_availability()
    print()

book3.rent_book()
print()

for book in books:
    print(f"Title: {book.title}")
    print(f"Author: {book.author}")
    book.check_availability()
    print()
```

```python
class Vehicle:
    def __init__(self, make, model, year, color):
        self.make = make
        self.model = model
        self.year = year
        self.color = color

    def display_info(self):
        print(f"Make: {self.make}")
        print(f"Model: {self.model}")
        print(f"Year: {self.year}")
        print(f"Color: {self.color}")

    def change_color(self, new_color):
        self.color = new_color
        print(f"The color of the {self.make} {self.model} has been changed to {self.color}.")

v1 = Vehicle("Toyota", "Camry", 2022, "Black")
v2 = Vehicle("Toyota", "Tacoma", 2019, "Blue")
v3 = Vehicle("Honda", "Civic", 1999, "Red")

for vehicle in [v1, v2, v3]:
    vehicle.display_info()
    print()

v2.change_color("White")
print()

v3.change_color("Green")
print()

for vehicle in [v1, v2, v3]:
    vehicle.display_info()
    print()
```
