```python
class Student:

    def __init__(self, name, major, credits):
        self.name = name
        self.major = major
        self.credits = credits

student1 = Student("Jordan", "Information Tech", "36")
student2 = Student("Kevin", "Cybersecurity", "60")
student3 = Student("Bob", "Data Science", "45")

print(f"Name: {student1.name}")
print(f"Major: {student1.major}")
print(f"Credits: {student1.credits}\n")

print(f"Name: {student2.name}")
print(f"Major: {student2.major}")
print(f"Credits: {student2.credits}\n")

print(f"Name: {student3.name}")
print(f"Major: {student3.major}")
print(f"Credits: {student3.credits}\n")
```

```python
class Product:

    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

product1 = Product("Laptop", "$1000", "7")

print(f"Name: {product1.name}")
print(f"Price: {product1.price}")
print(f"Quantity: {product1.quantity}\n")

product1.quantity = "5"

print(f"Name: {product1.name}")
print(f"Price: {product1.price}")
print(f"Quantity: {product1.quantity}")
```

```python
class Book:

    def __init__(self, title, author):
        self.title = title
        self.author = author

book1 = Book("The Hunger Games", "Suzanne Collins")
book2 = Book("The Great Gatsby", "F. Scott Fitzgerald")
book3 = Book("The Hobbit", "J.R.R. Tolkien")

for book in [book1, book2, book3]:
    print(f"Title: {book.title}")
    print(f"Author: {book.author}\n")
```

```python
class Movie:

    def __init__(self, title, director, release_date):
        self.title = title
        self.director = director
        self.release_date = release_date

movie1 = Movie("Pacific Rim", "Steven Soderbergh", "2013")
movie2 = Movie("Ford vs Ferrari", "James Mangold", "2019")
movie3 = Movie("Interstellar", "Christopher Nolan", "2014")

for movie in [movie1, movie2, movie3]:
    print(f"Title: {movie.title}")
    print(f"Director: {movie.director}")
    print(f"Release Date: {movie.release_date}\n")
```
