```python
class Book:
    def __init__(self, title, author, availability):
        self.title = title
        self.author = author
        self.availability = availability

    def display_info(self):
        print(f"Title: {self.title}")
        print(f"Author: {self.author}")
        if self.availability:
            print("Status: Available")
        else:
            print("Status: Checked Out")

    def check_out(self):
        if self.availability:
            self.availability = False
            print(f"{self.title} has been checked out.")
        else:
            print(f"{self.title} is not available for checkout.")

book1 = Book("The Hunger Games", "Suzanne Collins", True)
book2 = Book("To Kill a Mockingbird", "Harper Lee", True)
book3 = Book("The Great Gatsby", "F. Scott Fitzgerald", False)

print("Library Catalog\n")
for book in [book1, book2, book3]:
    book.display_info()
    print()

book2.check_out()
print()

print("Update Catalog\n")
for book in [book1, book2, book3]:
    book.display_info()
    print()
```
